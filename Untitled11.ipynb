```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
# Ensure the filename matches exactly what you uploaded to Colab
filename = 'merged_kick_balance_separate_trials.csv'

try:
    df = pd.read_csv(filename)
    print("File loaded successfully.")
except FileNotFoundError:
    print(f"Error: File '{filename}' not found. Please upload it to Colab.")

# 2. Identify Numeric Columns
# We automatically select columns with numeric data, excluding IDs and empty columns
numeric_cols = [col for col in df.select_dtypes(include=['number']).columns
                if col not in ['Participant_ID', 'Event', 'Unnamed: 0'] and 'Unnamed' not in col]

print(f"Analyzing {len(numeric_cols)} numeric columns...\n")

# 3. Perform Normality Tests (Shapiro-Wilk)
normality_results = []

for col in numeric_cols:
    # Drop missing values for the calculation
    data = df[col].dropna()

    # Shapiro-Wilk test requires at least 3 data points
    if len(data) >= 3:
        # Perform the test
        stat, p_value = stats.shapiro(data)

        # Calculate Skewness and Kurtosis for extra insight
        skew = stats.skew(data)
        kurt = stats.kurtosis(data)

        normality_results.append({
            'Column': col,
            'N': len(data),
            'Shapiro-Wilk Stat': round(stat, 4),
            'p-value': p_value,  # We keep full precision for logic, but you can round for display
            'Skewness': round(skew, 4),
            'Kurtosis': round(kurt, 4),
            'Is Normal (p>0.05)?': 'Yes' if p_value > 0.05 else 'No'
        })

# Create a DataFrame for the results
results_df = pd.DataFrame(normality_results)

# Display the table
print("--- Normality Test Results ---")
# pd.set_option('display.max_rows', None) # Uncomment to see all rows if truncated
print(results_df[['Column', 'N', 'p-value', 'Skewness', 'Is Normal (p>0.05)?']])


# 4. Visualization (Histograms & Q-Q Plots)
# We will plot the 4 most important columns to avoid cluttering the screen
# You can change 'cols_to_plot' to any column names from the list above
cols_to_plot = ['Max Total GRF',
                'COP path length [mm]_t1_left',
                'COP average velocity [mm/sec]_t1_left',
                'Resultant Velocity 1']

# Filter to keep only columns that actually exist in the file
cols_to_plot = [c for c in cols_to_plot if c in df.columns]

if cols_to_plot:
    plt.figure(figsize=(14, 4 * len(cols_to_plot)))

    for i, col in enumerate(cols_to_plot):
        data = df[col].dropna()

        # Histogram with KDE (kernel density estimate)
        plt.subplot(len(cols_to_plot), 2, 2*i + 1)
        sns.histplot(data, kde=True, color='skyblue')
        plt.title(f'Distribution: {col}')
        plt.xlabel('Value')

        # Q-Q Plot (Quantile-Quantile)
        plt.subplot(len(cols_to_plot), 2, 2*i + 2)
        stats.probplot(data, dist="norm", plot=plt)
        plt.title(f'Q-Q Plot: {col}')

    plt.tight_layout()
    plt.show()
else:
    print("None of the specified columns for plotting were found.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
from IPython.display import display, Markdown

# 1. Load Data
filename = 'merged_kick_balance_separate_trials.csv'
df = pd.read_csv(filename)

# 2. Filter for Shield Target
df_shield = df[df['Target'].astype(str).str.strip().str.lower() == 'shield'].copy()

# 3. Aggregate Data (Mean Max GRF per Participant per Condition)
# We take the mean to get one data point per participant
df_agg = df_shield.groupby(['Participant_ID', 'Technique', 'Side'])['Max Total GRF'].mean().reset_index()

print(f"Aggregated data for {df_agg['Participant_ID'].nunique()} participants.")

# 4. Statistical Analysis Function
def perform_grf_statistics(df_agg):
    techniques = ['front', 'middle', 'side']
    sides = ['left', 'right']

    # --- A. Compare Sides (Left vs Right) within each Technique ---
    side_results = []
    for tech in techniques:
        # Get paired data
        df_tech = df_agg[df_agg['Technique'] == tech]
        pivot = df_tech.pivot(index='Participant_ID', columns='Side', values='Max Total GRF').dropna()

        if len(pivot) > 0:
            stat, p = stats.wilcoxon(pivot['left'], pivot['right'])
            side_results.append({
                'Technique': tech.capitalize(),
                'Comparison': 'Left vs Right',
                'N': len(pivot),
                'p-value': p,
                'Significant': 'Yes' if p < 0.05 else 'No'
            })

    # --- B. Compare Techniques (Front vs Middle vs Side) for each Side ---
    tech_results = []
    for side in sides:
        # Get paired data across 3 techniques
        df_side = df_agg[df_agg['Side'] == side]
        pivot = df_side.pivot(index='Participant_ID', columns='Technique', values='Max Total GRF').dropna()

        # Ensure all columns exist
        cols = ['front', 'middle', 'side']
        if all(c in pivot.columns for c in cols):
            data = pivot[cols]

            # 1. Friedman Test
            stat, p_friedman = stats.friedmanchisquare(data['front'], data['middle'], data['side'])

            tech_results.append({
                'Side': side.capitalize(),
                'Comparison': 'All Techniques (Friedman)',
                'p-value': p_friedman,
                'Significant': 'Yes' if p_friedman < 0.05 else 'No'
            })

            # 2. Post-hoc Wilcoxon
            if p_friedman < 0.05:
                pairs = [('front', 'middle'), ('middle', 'side'), ('front', 'side')]
                for t1, t2 in pairs:
                    stat, p_wilc = stats.wilcoxon(data[t1], data[t2])
                    p_corr = min(p_wilc * 3, 1.0) # Bonferroni

                    tech_results.append({
                        'Side': side.capitalize(),
                        'Comparison': f'{t1.capitalize()} vs {t2.capitalize()}',
                        'p-value': p_corr,
                        'Significant': 'Yes' if p_corr < 0.05 else 'No'
                    })

    return pd.DataFrame(side_results), pd.DataFrame(tech_results)

# 5. Run Analysis
df_side_stats, df_tech_stats = perform_grf_statistics(df_agg)

# 6. Plot and Show Results
sns.set_style("whitegrid")
plt.figure(figsize=(10, 6))

sns.boxplot(data=df_agg, x='Technique', y='Max Total GRF', hue='Side', palette='Set2', order=['front', 'middle', 'side'])
plt.title('Max Total GRF by Technique and Side (Aggregated Mean per Participant)', fontsize=15)
plt.ylabel('Mean Max Total GRF (N)')
plt.xlabel('Technique')
plt.legend(title='Side')
plt.tight_layout()
plt.show()

display(Markdown("### 1. Differences Between Sides (Left vs Right)"))
display(df_side_stats)

display(Markdown("### 2. Differences Between Techniques (Front vs Middle vs Side)"))
display(df_tech_stats)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
from IPython.display import display, Markdown

# 1. Load Data
filename = 'merged_kick_balance_separate_trials.csv'
df = pd.read_csv(filename)

# 2. Filter for Shield Target and Unique Participants
# We only want one row per participant for balance analysis
df_shield = df[df['Target'].astype(str).str.strip().str.lower() == 'shield'].copy()
balance_cols = [c for c in df.columns if 'COP' in c]
df_unique = df_shield[['Participant_ID'] + balance_cols].drop_duplicates(subset=['Participant_ID'])

print(f"Analyzing balance data for {len(df_unique)} unique participants.")

# 3. Define Metrics
metrics = [
    {'name': 'COP average velocity [mm/sec]', 'short': 'Velocity', 'unit': 'mm/s'},
    {'name': 'COP path length [mm]', 'short': 'Path Length', 'unit': 'mm'}
]
trials = ['t1', 't2', 't3']
sides = ['left', 'right']

# 4. Statistical Analysis Function
def perform_statistics(df, metric_info):
    base_name = metric_info['name']
    short_name = metric_info['short']

    # --- A. Compare Sides (Left vs Right) within each Trial ---
    side_results = []
    for t in trials:
        col_l = f"{base_name}_{t}_left"
        col_r = f"{base_name}_{t}_right"

        # Drop NaNs to ensure paired samples
        valid_data = df[[col_l, col_r]].dropna()

        if len(valid_data) > 0:
            stat, p = stats.wilcoxon(valid_data[col_l], valid_data[col_r])
            side_results.append({
                'Trial': t.upper(),
                'Comparison': 'Left vs Right',
                'N': len(valid_data),
                'p-value': p,
                'Significant (p<0.05)': 'Yes' if p < 0.05 else 'No'
            })

    # --- B. Compare Trials (t1 vs t2 vs t3) for each Side ---
    trial_results = []
    for side in sides:
        # Columns for this side across 3 trials
        cols = [f"{base_name}_{t}_{side}" for t in trials]
        valid_data = df[cols].dropna()

        if len(valid_data) > 0:
            # 1. Friedman Test (Global difference)
            stat, p_friedman = stats.friedmanchisquare(valid_data[cols[0]], valid_data[cols[1]], valid_data[cols[2]])

            trial_results.append({
                'Side': side.capitalize(),
                'Comparison': 'All Trials (Friedman)',
                'p-value': p_friedman,
                'Significant': 'Yes' if p_friedman < 0.05 else 'No'
            })

            # 2. Post-hoc Wilcoxon (if Friedman is significant)
            if p_friedman < 0.05:
                pairs = [(0, 1), (1, 2), (0, 2)] # (t1,t2), (t2,t3), (t1,t3)
                for i, j in pairs:
                    c1, c2 = cols[i], cols[j]
                    stat, p_wilc = stats.wilcoxon(valid_data[c1], valid_data[c2])

                    # Bonferroni Correction (multiply p by number of comparisons, i.e., 3)
                    p_corr = min(p_wilc * 3, 1.0)

                    trial_results.append({
                        'Side': side.capitalize(),
                        'Comparison': f'{trials[i].upper()} vs {trials[j].upper()}',
                        'p-value': p_corr,
                        'Significant': 'Yes' if p_corr < 0.05 else 'No'
                    })

    return pd.DataFrame(side_results), pd.DataFrame(trial_results)

# 5. Generate Plots and Tables
sns.set_style("whitegrid")

for metric in metrics:
    display(Markdown(f"# --- Analysis: {metric['short']} ---"))

    # Run Statistics
    df_side_stats, df_trial_stats = perform_statistics(df_unique, metric)

    # Plot Each Trial (Left vs Right)
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    fig.suptitle(f"{metric['short']} - Comparison by Trial", fontsize=16)

    for i, t in enumerate(trials):
        # Prepare Data
        col_l = f"{metric['name']}_{t}_left"
        col_r = f"{metric['name']}_{t}_right"
        plot_data = df_unique[[col_l, col_r]].rename(columns={col_l: 'Left', col_r: 'Right'})
        df_melt = plot_data.melt(var_name='Side', value_name='Value')

        # Plot
        sns.boxplot(data=df_melt, x='Side', y='Value', ax=axes[i], palette='Set2')
        axes[i].set_title(f"Trial {t.upper()}")
        axes[i].set_ylabel(metric['unit'])
        axes[i].set_xlabel('')

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

    # Display Statistics Tables
    display(Markdown("### 1. Differences Between Sides (Left vs Right)"))
    display(df_side_stats)

    display(Markdown("### 2. Differences Between Trials (t1 vs t2 vs t3)"))
    display(df_trial_stats)
    print("\n" + "="*100 + "\n")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# 1. Load Data
filename = 'merged_kick_balance_separate_trials.csv'
df = pd.read_csv(filename)

# 2. Clean Column Names (Remove extra spaces)
df.columns = df.columns.str.strip()

# 3. Filter for Shield Target Only
df_shield = df[df['Target'].astype(str).str.strip().str.lower() == 'shield'].copy()

# 4. Define Metrics
# Kick Metrics (Rows)
kick_metrics = ['Max Total GRF'] + \
               [f'Resultant Velocity {i}' for i in range(1, 7)] + \
               [f'Resultant Acceleration {i}' for i in range(1, 7)]

# Balance Metrics (Columns) - All COP columns
balance_metrics = [col for col in df.columns if 'COP' in col]

# Convert to Numeric
for col in kick_metrics + balance_metrics:
    df_shield[col] = pd.to_numeric(df_shield[col], errors='coerce')

# 5. Analysis Loop
techniques = ['front', 'middle', 'side']
sides = ['left', 'right']

sns.set_style("white")

for tech in techniques:
    for side in sides:
        # Filter Data Subset
        subset = df_shield[(df_shield['Technique'] == tech) & (df_shield['Side'] == side)]

        if subset.empty:
            continue

        # Calculate Spearman Correlation
        full_corr = subset[kick_metrics + balance_metrics].corr(method='spearman')

        # Slice to keep only Kick vs Balance
        corr_matrix = full_corr.loc[kick_metrics, balance_metrics]

        # --- A. Display Heatmap ---
        plt.figure(figsize=(14, 10))
        sns.heatmap(corr_matrix,
                    annot=True,
                    fmt=".2f",
                    cmap='coolwarm',
                    center=0,
                    vmin=-1, vmax=1,
                    linewidths=0.5,
                    annot_kws={"size": 8},
                    cbar_kws={'label': 'Spearman Correlation'})

        plt.title(f'Spearman Correlation: Kick vs Balance\n({tech.capitalize()} Kick - {side.capitalize()} Side)', fontsize=16)
        plt.ylabel('Kick Performance Metrics', fontsize=12)
        plt.xlabel('Balance Metrics (COP)', fontsize=12)
        plt.xticks(rotation=45, ha='right')
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show() # Show plot in notebook

        # --- B. Display Correlation Table ---
        display(Markdown(f"### Correlation Table: {tech.capitalize()} Kick - {side.capitalize()} Side"))
        display(corr_matrix.round(2))
        print("\n" + "="*100 + "\n") # Separator line
```
